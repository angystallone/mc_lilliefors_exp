```python
import numpy as np
from mc_lilliefors_exp import estimate_mc_lilliefors
```

```python
# Simulate binned magnitudes by sampling from the geometric distribution

def simulate_gr_catalog(NumEvents, b_value, MagnMin, delta_m):

    beta = b_value * np.log(10)
    p = 1 - np.exp(-beta * delta_m)
    k = np.random.geometric(p, size=NumEvents) - 1

    return MagnMin + k * delta_m 
```

```python
# Synthetic catalog params (ground truth)

Mc_true = 1.0
b_value = 1.0
delta_m = 0.1
n_events = 1_000_000
```

```python
# TEST PARAMS

# Significance level for Lilliefors test
alpha = 0.05 
# Number of independent dithering realizations
n_dithers = 100 

# To assess robustness against b-value misspecification, choose a value
# different from the true b-value used to generate the synthetic catalog
b_value_user = 0.9
```

```python
np.random.seed(123)
print("Generating synthetic catalog")
mags = simulate_gr_catalog(
    NumEvents=n_events,
    b_value=b_value,
    MagnMin=Mc_true,
    delta_m=delta_m
)
print("Magn Min = ", min(mags))
```

```python
print("Estimating Mc using Lilliefors with exponential dithering")

if b_value_user == b_value:
    print("The supplied b-value matches the underlying theoretical exponential slope")
else:
    print("The supplied b-value differs from the underlying theoretical exponential slope")

np.random.seed(456)
mc_est, pvals = estimate_mc_lilliefors(
    magnitudes=mags,
    delta_m=delta_m,
    b_value=b_value_user,
    alpha=alpha,
    n_dithers=n_dithers,
    min_events=50
)

print("--- RESULTS ---")
print(f"True Mc: {Mc_true}")
print(f"Estimated Mc: {mc_est}")
print("Mean p-values by candidate Mc:")
for m, p in sorted(pvals.items()):
    print(f"  Mc = {m:.1f}: p = {p:.3f}")

if mc_est is None:
    print("WARNING: estimator returned None (no stable Mc)")

if np.isclose(mc_est, Mc_true):
    print("TEST PASSED: estimator is equal to the true Mc")
else:
    print("TEST FAILED: estimator is too far from true Mc")
```
